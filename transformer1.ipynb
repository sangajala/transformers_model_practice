# Sentiment Analysis with Hugging Face Transformers
This notebook demonstrates how to perform sentiment analysis on customer reviews using a pre-trained model from Hugging Face, running entirely in Google Colab without requiring an API key.

```python
# Install the Hugging Face Transformers library
!pip install -q transformers
```

```python
from transformers import pipeline

# Initialize a sentiment analysis pipeline using a small, public model
# distilbert-base-uncased-finetuned-sst-2-english is highly accurate and lightweight
sentiment_pipeline = pipeline("sentiment-analysis", model="distilbert-base-uncased-finetuned-sst-2-english")
```

```python
# Define 5 sample customer reviews
reviews = [
    "This product is absolutely amazing! It exceeded all my expectations and works perfectly.",
    "Very disappointed with the quality. It broke within the first two days of usage.",
    "It is okay, nothing special. It does the job but has some minor issues.",
    "The customer service team was extremely helpful and resolved my query in minutes.",
    "I hate the design and it is far too expensive for what it offers."
]

# Run sentiment analysis
results = sentiment_pipeline(reviews)

# Print the results with labels and confidence scores
for i, (review, result) in enumerate(zip(reviews, results), 1):
    print(f"Review {i}: '{review}'")
    print(f"Sentiment: {result['label']} | Confidence Score: {result['score']:.4f}\n")
```

## Understanding the Attention Mechanism

Transformers rely on the **Attention Mechanism** to process text. Here are two questions explaining how it works:

### Question 1: What is 'Self-Attention' and how does it help the model understand context?
**Answer:** Self-attention allows a model to look at other words in the input sequence to help decode a specific word. For example, in the sentence *'The bank of the river was muddy'*, self-attention helps the model associate the word *'bank'* with *'river'* (and thus water) rather than a financial institution, by dynamically weighting the relationships between all words in the sentence.

### Question 2: Why are 'Attention Weights' or 'Scores' important during sentiment analysis?
**Answer:** Attention weights represent how much focus the model places on different words when making its final prediction. In sentiment analysis, words like *'amazing'*, *'disappointed'*, or *'hate'* will receive much higher attention scores than neutral words like *'this'* or *'the'*, directly guiding the model to output a Positive or Negative label.
